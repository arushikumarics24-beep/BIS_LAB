```python

import random

# -----------------------------------
# Ant Colony Optimization Algorithm
# -----------------------------------

def ant_colony_optimization(
    tables,
    distance,
    num_ants=10,
    max_iterations=100,
    alpha=1,
    beta=2,
    rho=0.5,
    Q=100
):

    n = len(tables)

    # Initialize pheromone matrix
    pheromone = [[1.0 for j in range(n)] for i in range(n)]

    best_route = None
    best_distance = float('inf')

    # Calculate initial route cost
    initial_route = list(range(n)) + [0]

    initial_distance = sum(
        distance[initial_route[i]][initial_route[i + 1]]
        for i in range(n)
    )

    # Main iterations
    for iteration in range(max_iterations):

        all_routes = []

        # Generate solutions for each ant
        for ant in range(num_ants):

            # Select a random starting city
            start = random.randint(0, n - 1)

            route = [start]
            unvisited = list(range(n))
            unvisited.remove(start)

            # Visit all remaining cities
            while unvisited:

                current = route[-1]

                # Calculate selection probabilities
                probabilities = []

                for city in unvisited:

                    tau = pheromone[current][city]
                    eta = 1 / (distance[current][city] + 0.0001)

                    probability = (tau ** alpha) * (eta ** beta)

                    probabilities.append(probability)

                # Normalize probabilities
                total = sum(probabilities)

                probabilities = [
                    p / total for p in probabilities
                ]

                # Select next city
                next_city = random.choices(
                    unvisited,
                    weights=probabilities,
                    k=1
                )[0]

                route.append(next_city)
                unvisited.remove(next_city)

            # Return to starting city
            route.append(start)

            # Calculate total route distance
            route_distance = sum(
                distance[route[i]][route[i + 1]]
                for i in range(n)
            )

            all_routes.append((route, route_distance))

            # Update best route
            if route_distance < best_distance:

                best_distance = route_distance
                best_route = route.copy()

        # Evaporate pheromone on all paths
        for i in range(n):
            for j in range(n):

                pheromone[i][j] *= (1 - rho)

        # Deposit pheromone on ant paths
        for route, route_distance in all_routes:

            deposit = Q / route_distance

            for i in range(n):

                a = route[i]
                b = route[i + 1]

                pheromone[a][b] += deposit
                pheromone[b][a] += deposit

    return best_route, best_distance, initial_distance


# -----------------------------------
# Main Program
# -----------------------------------

# Database tables
tables = ["Table1", "Table2", "Table3", "Table4"]

# Distance matrix representing
# estimated network transmission cost (MB)

distance = [
    [0, 80, 70, 10],
    [80, 0, 10, 70],
    [70, 10, 0, 80],
    [10, 70, 80, 0]
]

# Execute ACO
best_route, best_distance, initial_distance = (
    ant_colony_optimization(
        tables,
        distance,
        num_ants=10,
        max_iterations=100,
        alpha=1,
        beta=2,
        rho=0.5,
        Q=100
    )
)

# Display results
print("\n------ Execution Results ------")

print("\nOptimal Query Execution Plan:")

for i in range(len(best_route) - 1):
    print(tables[best_route[i]], end="")

    if i < len(best_route) - 2:
        print(" -> ", end="")

print("\n")

print("Minimum Network Transmission Cost:",
      best_distance, "MB")

print("Initial Network Transmission Cost:",
      initial_distance, "MB")

print("Network Cost Saved:",
      initial_distance - best_distance, "MB")
```
